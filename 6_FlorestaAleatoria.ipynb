```python
import pandas as pd
import numpy as np
import joblib
import matplotlib.pyplot as plt

from sklearn.base import clone
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import (
    StratifiedKFold,
    RandomizedSearchCV,
    cross_val_score
)
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from imblearn.pipeline import Pipeline
from imblearn.under_sampling import RandomUnderSampler
```

```python
# Define nome da base
base_name = "Wine"
```

```python
#===============================================
# Define caminho padrão
base_path = "/content/drive/MyDrive/Trabalho_IA/Arquivos/"

# Define colunas para remover
remove_pos_col = [-1]

# Define categorias da base
base_categorias = [-1]

#===============================================
# Lê dados de treinamento completo sem escalonamento
dados_treinamento_completo_originais = pd.read_csv(
    base_path + base_name + "/" + "Dados_" + base_name + "_Treinamento_Completo_Originais.csv",
    sep=";",
    encoding="utf-8-sig",
    low_memory=False
)

X_1 = dados_treinamento_completo_originais

# Remove colunas
X_treinamento_completo_original = X_1.drop(X_1.columns[remove_pos_col], axis=1)
Y_treinamento_completo_original = X_1.iloc[:, base_categorias].values

X_train_cv = X_treinamento_completo_original
Y_train_cv = Y_treinamento_completo_original.ravel()

# Mostra dados do arquivo
print("Dados do conjunto Treinamento Completo sem escalonamento\n")
display(X_1)

'''
#===============================================
# Lê dados de treinamento completo com escalonamento
dados_treinamento_completo_scaled = pd.read_csv(
    base_path + base_name + "/" + "Dados_" + base_name + "_Treinamento_Completo_Escalonados.csv",
    sep=";",
    encoding="utf-8-sig",
    low_memory=False
)

X_1 = dados_treinamento_completo_scaled

# Remove colunas
X_treinamento_completo_scaled = X_1.drop(X_1.columns[remove_pos_col], axis=1)
Y_treinamento_completo_scaled = X_1.iloc[:, base_categorias].values

# Mostra dados do arquivo
print("Dados do conjunto Treinamento Completo com escalonamento\n")
display(X_1)

'''

'''
#===============================================
# Lê dados de treinamento sem escalonamento
dados_train_original = pd.read_csv(
    base_path + base_name + "/" + "Dados_" + base_name + "_Treinamento_Originais.csv",
    sep=";",
    encoding="utf-8-sig",
    low_memory=False
)

X_1 = dados_train_original

# Remove colunas
X_train_original = X_1.drop(X_1.columns[remove_pos_col], axis=1)
Y_train_original = X_1.iloc[:, base_categorias].values

# Mostra dados do arquivo
print("Dados do conjunto Treinamento sem escalonamento\n")
display(X_1)

'''

#===============================================
# Lê dados de treinamento com escalonamento
dados_train_scaled = pd.read_csv(
    base_path + base_name + "/" + "Dados_" + base_name + "_Treinamento_Escalonados.csv",
    sep=";",
    encoding="utf-8-sig",
    low_memory=False
)

X_1 = dados_train_scaled

# Remove colunas
X_train_scaled = X_1.drop(X_1.columns[remove_pos_col], axis=1)
Y_train_scaled = X_1.iloc[:, base_categorias].values

X_train = X_train_scaled
Y_train = Y_train_scaled

# Mostra dados do arquivo
print("Dados do conjunto Treinamento com escalonamento\n")
display(X_1)

#===============================================
# Lê dados de teste sem escalonamento
dados_test_original = pd.read_csv(
    base_path + base_name + "/" + "Dados_" + base_name + "_Teste_Originais.csv",
    sep=";",
    encoding="utf-8-sig",
    low_memory=False
)

X_1 = dados_test_original

# Remove colunas
X_test_original = X_1.drop(X_1.columns[remove_pos_col], axis=1)
Y_test_original = X_1.iloc[:, base_categorias].values

# Mostra dados do arquivo
print("Dados do conjunto Teste sem escalonamento\n")
display(X_1)

#===============================================
# Lê dados de teste com escalonamento
dados_test_scaled = pd.read_csv(
    base_path + base_name + "/" + "Dados_" + base_name + "_Teste_Escalonados.csv",
    sep=";",
    encoding="utf-8-sig",
    low_memory=False
)

X_1 = dados_test_scaled

# Remove colunas
X_test_scaled = X_1.drop(X_1.columns[remove_pos_col], axis=1)
Y_test_scaled = X_1.iloc[:, base_categorias].values

X_test = X_test_scaled
Y_test = Y_test_scaled

# Mostra dados do arquivo
print("Dados do conjunto Teste com escalonamento\n")
display(X_1)

#===============================================
# Cria o modelo Floresta Aleatória
modelo_rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

# Treina o modelo
modelo_rf.fit(X_train, np.asarray(Y_train).ravel())

print("\nFloresta Aleatória treinada com sucesso!\n")

#===============================================
# Realiza as previsões no conjunto de teste
Y_pred_rf = modelo_rf.predict(X_test)

#===============================================
# Calcula as métricas
print("Acurácia:", accuracy_score(Y_test, Y_pred_rf))
print("Precisão (classe 1):", precision_score(Y_test, Y_pred_rf, pos_label=1, zero_division=0))
print("Revocação (classe 1):", recall_score(Y_test, Y_pred_rf, pos_label=1, zero_division=0))
print("F1-score (classe 1):", f1_score(Y_test, Y_pred_rf, pos_label=1, zero_division=0))
print("F1-macro:", f1_score(Y_test, Y_pred_rf, average="macro", zero_division=0))

# Relatório completo
print("\nRelatório de classificação:\n")
print(classification_report(
    Y_test,
    Y_pred_rf,
    digits=4,
    zero_division=0
))

# Matriz de confusão
print("\nMatriz de confusão:\n")
print(confusion_matrix(Y_test, Y_pred_rf))
print("\n")

#===============================================
# Recupera o conjunto Treinamento Completo sem escalonamento, antes do balanceamento
print("Treinamento completo para validação cruzada:", X_train_cv.shape)
print("Distribuição das classes:")
print(pd.Series(np.asarray(Y_train_cv).ravel()).value_counts())

# ===============================================
# Define a validação cruzada
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# ===============================================
# Recupera o escalonador utilizado no prepara_dados
scaler = joblib.load(
    base_path + base_name + "/Scaler_" + base_name + ".pkl"
)

print(
    "Escalonador utilizado:",
    type(scaler).__name__ if scaler is not None else "Nenhum"
)

# ===============================================
# Calcula a proporção original das classes
classes, quantidades = np.unique(
    np.asarray(Y_train_cv).ravel(),
    return_counts=True
)

proporcao = quantidades.min() / quantidades.max()

print("Proporção entre classes:", round(proporcao, 4))

# ===============================================
# Define as configurações de balanceamento
parametros_balanceamento = [
    {"balanceador": ["passthrough"]}
]

if len(classes) == 2 and proporcao < 0.5:
    parametros_balanceamento = [
        {
            "balanceador": [RandomUnderSampler(random_state=42)],
            "balanceador__sampling_strategy": [0.5]
        }
    ]

print("Configurações de balanceamento:", parametros_balanceamento)

# ===============================================
# MODELO E HIPERPARÂMETROS DA FLORESTA ALEATÓRIA
modelo = RandomForestClassifier(
    random_state=42,
    n_jobs=1
)

# Seleção automática da grade conforme o tamanho da base
if len(X_train_cv) <= 10000:
    # Bases menores: 225 combinações possíveis
    parametros_modelo = {
        "modelo__n_estimators": [50, 100, 200, 300, 500],
        "modelo__max_depth": [5, 10, 15, 20, None],
        "modelo__max_features": ["sqrt", "log2", None],
        "modelo__min_samples_leaf": [1, 2, 5]
    }
    n_iter_busca = 30
else:
    # Bases maiores: 96 combinações possíveis
    parametros_modelo = {
        "modelo__n_estimators": [50, 100, 200, 300],
        "modelo__max_depth": [5, 10, 20, None],
        "modelo__max_features": ["sqrt", "log2"],
        "modelo__min_samples_leaf": [1, 5, 10]
    }
    n_iter_busca = 20

print("Grade Floresta Aleatória:", parametros_modelo)

# ===============================================
# Define o pipeline
pipeline = Pipeline([
    ("balanceador", "passthrough"),
    ("scaler", clone(scaler) if scaler is not None else "passthrough"),
    ("modelo", modelo)
])

# ===============================================
# Combina os hiperparâmetros do modelo e do balanceamento
parametros = [
    {**modelo_parametros, **opcao}
    for modelo_parametros in (
        parametros_modelo
        if isinstance(parametros_modelo, list)
        else [parametros_modelo]
    )
    for opcao in parametros_balanceamento
]

# ===============================================
# Configura a busca
grid = RandomizedSearchCV(
    estimator=pipeline,
    param_distributions=parametros,
    n_iter=n_iter_busca,
    scoring="f1_macro",
    cv=cv,
    random_state=42,
    n_jobs=-1,
    error_score="raise"
)

# Executa a busca somente no treinamento completo
grid.fit(X_train_cv, np.asarray(Y_train_cv).ravel())

print("\nMelhores parâmetros:", grid.best_params_)
print("Melhor F1-macro na validação:", grid.best_score_)

# ===============================================
# Recupera o melhor modelo
modelo_otimizado = grid.best_estimator_

# ===============================================
# Avalia a saturação conforme o número de árvores
# Mantém os demais hiperparâmetros do melhor modelo
numeros_arvores = [10, 25, 50, 100, 200, 300, 500]
resultados_f1 = []

for n_arvores in numeros_arvores:

    modelo_saturacao = clone(modelo_otimizado)
    modelo_saturacao.set_params(
        modelo__n_estimators=n_arvores
    )

    # Avalia somente no treinamento por validação cruzada
    scores = cross_val_score(
        modelo_saturacao,
        X_train_cv,
        np.asarray(Y_train_cv).ravel(),
        scoring="f1_macro",
        cv=cv,
        n_jobs=-1
    )

    resultados_f1.append(scores.mean())

    print(
        f"Número de árvores: {n_arvores} | "
        f"F1-macro validação: {scores.mean():.4f}"
    )

# ===============================================
# Constrói o gráfico de saturação
plt.figure(figsize=(9, 5))

plt.plot(
    numeros_arvores,
    resultados_f1,
    marker="o"
)

plt.xlabel("Número de árvores (n_estimators)")
plt.ylabel("F1-macro na validação cruzada")
plt.title("Floresta Aleatória - Saturação do desempenho")
plt.grid(True)
plt.show()

# ===============================================
# Realiza as previsões
# O pipeline aplica automaticamente o escalonamento
Y_pred_treino = modelo_otimizado.predict(X_train_cv)
Y_pred_otimizado = modelo_otimizado.predict(X_test_original)

# ===============================================
# Avalia o desempenho no treinamento
f1_treino = f1_score(
    np.asarray(Y_train_cv).ravel(),
    Y_pred_treino,
    average="macro",
    zero_division=0
)

f1_teste = f1_score(
    np.asarray(Y_test_original).ravel(),
    Y_pred_otimizado,
    average="macro",
    zero_division=0
)

print("\nComparação de desempenho:")
print(f"F1-macro treino: {f1_treino:.4f}")
print(f"F1-macro teste:  {f1_teste:.4f}")
print(f"Diferença:      {f1_treino - f1_teste:.4f}")

print("\nRelatório de classificação - TREINO:\n")
print(classification_report(
    np.asarray(Y_train_cv).ravel(),
    Y_pred_treino,
    digits=4,
    zero_division=0
))

print("\nMatriz de confusão - TREINO:\n")
print(confusion_matrix(
    np.asarray(Y_train_cv).ravel(),
    Y_pred_treino
))

# ===============================================
# Avalia o modelo no teste
print("\nRelatório de classificação pós validação cruzada:\n")
print(classification_report(
    np.asarray(Y_test_original).ravel(),
    Y_pred_otimizado,
    digits=4,
    zero_division=0
))

print("\nMatriz de confusão pós validação cruzada:\n")
print(confusion_matrix(
    np.asarray(Y_test_original).ravel(),
    Y_pred_otimizado
))
print("\n")
```
